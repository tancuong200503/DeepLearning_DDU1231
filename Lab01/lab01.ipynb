**Yêu cầu:** Cho biết kích thước, số chiều và kiểu dữ liệu của Tensor `a`.

```python
import torch

a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.int32
)

print(a)
print("Kích thước:", a.shape)
print("Số chiều:", a.ndim)
print("Kiểu dữ liệu:", a.dtype)
```

### Bài 2. Tạo các Tensor thường dùng

### Bài 4. Slicing và Indexing

Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

**Yêu cầu:**
- Cho biết shape của `a[1]` và `a[1:, 2:4]`.
- Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

```python
import torch

a = torch.randn(3, 4, 5)

print("Original tensor:")
print(a)

print("\nTensor a[1]:")
print(a[1])

print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])

print("\nShape of a[1]:", a[1].shape)
print("Shape of a[1:, 2:4]:", a[1:, 2:4].shape)

print("\nHai phần tử cuối trên chiều thứ ba:")
print(a[:, :, -2:])
```

#### Nhận xét

Tensor ban đầu `a` có kích thước `3 × 4 × 5`.

- `a[1]` lấy phần tử có chỉ số 1 trên chiều thứ nhất, tức khối thứ hai của Tensor. Kết quả có shape `torch.Size([4, 5])`.
- `a[1:, 2:4]` lấy các phần tử từ chỉ số 1 trở đi trên chiều thứ nhất và từ chỉ số 2 đến trước chỉ số 4 trên chiều thứ hai. Kết quả có shape `torch.Size([2, 2, 5])`.
- `a[:, :, -2:]` giữ nguyên hai chiều đầu và lấy hai phần tử cuối cùng trên chiều thứ ba. Kết quả có shape `torch.Size([3, 4, 2])`.

Qua đó, indexing dùng để truy xuất phần tử hoặc một phần cụ thể của Tensor, còn slicing cho phép lấy một phạm vi dữ liệu theo từng chiều.

### Bài 5. Reshape và Flatten

Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

**Yêu cầu:**
- Kiểm tra số phần tử trước và sau khi reshape.
- Thử reshape Tensor thành kích thước 2 × 6.

```python
import torch

a = torch.randn(3, 4)

print("Original:")
print(a)

print("\nFlatten:")
print(a.ravel())

print("\nReshape to 3 x 2 x 2:")
b = a.reshape(3, 2, 2)
print(b)

print("\nSố phần tử ban đầu:", a.numel())
print("Số phần tử sau khi reshape:", b.numel())

print("\nReshape to 2 x 6:")
c = a.reshape(2, 6)
print(c)

print("\nShape ban đầu:", a.shape)
print("Shape sau khi reshape 3 x 2 x 2:", b.shape)
print("Shape sau khi reshape 2 x 6:", c.shape)
```

#### Nhận xét

Tensor ban đầu có kích thước `3 × 4`, tương ứng với **12 phần tử**.

- `a.ravel()` chuyển Tensor thành Tensor một chiều có 12 phần tử.
- `a.reshape(3, 2, 2)` thay đổi Tensor thành kích thước `3 × 2 × 2` nhưng vẫn giữ nguyên 12 phần tử.
- `a.reshape(2, 6)` thay đổi Tensor thành kích thước `2 × 6` và cũng giữ nguyên 12 phần tử.
- `numel()` cho biết tổng số phần tử của Tensor.

Như vậy, thao tác reshape chỉ thay đổi hình dạng của Tensor mà không làm thay đổi tổng số phần tử.

### Bài 6. Phép toán và nhân ma trận

Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử `@`.

**Yêu cầu:**
- Phân biệt phép toán `a * b` với `a @ b.T`.
- Cho biết shape của kết quả `a @ b.T`.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)

print("\nb:")
print(b)

print("\na * b:")
print(a * b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication:")
print(a @ b.T)

print("\nShape của a * b:", (a * b).shape)
print("Shape của a @ b.T:", (a @ b.T).shape)
```

#### Nhận xét

- `a * b` thực hiện phép nhân theo từng phần tử tương ứng của hai Tensor. Vì `a` và `b` đều có kích thước `2 × 3`, kết quả cũng có shape `torch.Size([2, 3])`.
- `a @ b.T` thực hiện phép nhân ma trận. Tensor `b` có kích thước `2 × 3`, sau khi chuyển vị `b.T` có kích thước `3 × 2`.
- Do đó, phép nhân giữa `a` có kích thước `2 × 3` và `b.T` có kích thước `3 × 2` tạo ra Tensor có shape `torch.Size([2, 2])`.
- Như vậy, toán tử `*` dùng để nhân theo từng phần tử, còn toán tử `@` dùng để thực hiện phép nhân ma trận.

### Bài 7. Các hàm thống kê trên Tensor

Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

**Yêu cầu:** Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.

```python
import torch

a = torch.randn(3, 4)

print("Tensor a:")
print(a)

print("\n--- dim = 0 ---")

print("\nMean:")
print(torch.mean(a, dim=0))

print("\nStandard deviation:")
print(torch.std(a, dim=0))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

print("\n--- dim = 1 ---")

print("\nMean:")
print(torch.mean(a, dim=1))

print("\nStandard deviation:")
print(torch.std(a, dim=1))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

#### Nhận xét

- Khi `dim=0`, các phép toán được thực hiện theo chiều thứ nhất của Tensor, tức là tính theo từng cột. Với Tensor kích thước `3 × 4`, kết quả của `mean` và `std` có 4 phần tử.
- Khi `dim=1`, các phép toán được thực hiện theo chiều thứ hai, tức là tính theo từng hàng. Kết quả của `mean` và `std` có 3 phần tử.
- `torch.cumsum()` tính tổng tích lũy theo chiều được chỉ định. Với `dim=0`, tổng được tích lũy theo các hàng cho từng cột; với `dim=1`, tổng được tích lũy theo các cột trong từng hàng.

Như vậy, thay đổi giá trị `dim` sẽ làm thay đổi chiều mà phép toán thống kê được thực hiện.

### Bài 8. Autograd cơ bản

Tạo biến có `requires_grad=True` và sử dụng `backward()` để tính đạo hàm tự động.

**Yêu cầu:**
- Tính thủ công đạo hàm của `y = x²` tại `x = 3.6` và so sánh với `x.grad`.
- Đổi hàm thành `y = 3*x**2 + 2*x + 1` và kiểm tra gradient.

```python
import torch

# Trường hợp 1: y = x^2
x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x
y.backward()

print("--- Trường hợp 1: y = x^2 ---")
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)


# Trường hợp 2: y = 3*x^2 + 2*x + 1
x = torch.tensor(
    3.6,
    requires_grad=True
)

y = 3 * x**2 + 2 * x + 1
y.backward()

print("\n--- Trường hợp 2: y = 3*x^2 + 2*x + 1 ---")
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

#### Nhận xét

Với hàm `y = x²`, đạo hàm theo `x` là: `dy/dx = 2x`

Tại `x = 3.6`: `dy/dx = 2 × 3.6 = 7.2`

Kết quả này trùng với giá trị `x.grad` do PyTorch tính bằng `backward()`.

Khi đổi hàm thành `y = 3*x**2 + 2*x + 1`, đạo hàm là: `dy/dx = 6x + 2`

Tại `x = 3.6`: `dy/dx = 6 × 3.6 + 2 = 23.6`

Giá trị này cũng trùng với gradient được PyTorch tính tự động. Qua đó, `autograd` cho phép PyTorch theo dõi các phép toán và tự động tính gradient của Tensor có `requires_grad=True`.

### Bài 9. Gradient Descent tìm hệ số đa thức

Sinh dữ liệu từ đa thức `y = x² + 2x + 3`, sau đó sử dụng NAdam và autograd để tìm lại ba hệ số.

**Yêu cầu:**
- So sánh ba hệ số học được với `[1, 2, 3]`.
- Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()` và `optimizer.step()`.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()

    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))

    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

#### Nhận xét

Ba hệ số ban đầu được khởi tạo ngẫu nhiên là `[0.6880, 0.0690, 1.4652]`. Sau 1000 epoch, mô hình học được các hệ số xấp xỉ `[1.0001, 2.0003, 2.9938]`, rất gần với các hệ số thực `[1, 2, 3]` của đa thức `y = x² + 2x + 3`.

Trong quá trình tối ưu:
- `optimizer.zero_grad()` dùng để xóa gradient của vòng lặp trước, tránh gradient bị cộng dồn.
- `mse.backward()` thực hiện lan truyền ngược để tính gradient của hàm mất mát MSE đối với các hệ số `w`.
- `optimizer.step()` sử dụng các gradient đã tính để cập nhật các hệ số `w` nhằm làm giảm sai số.

Như vậy, thông qua autograd và optimizer NAdam, PyTorch có thể tối ưu các tham số và tìm lại gần đúng các hệ số của đa thức ban đầu.

### Bài 10. Giải hệ phương trình bằng Autograd

Tối thiểu hóa tổng sai số bình phương để tìm `A`, `B`, `C`, `D` thỏa bốn phương trình.

**Yêu cầu:**
- Kiểm tra nghiệm thu được với bốn phương trình: `A+B=9`, `C-D=1`, `A+C=8`, `B-D=2`.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("\nKiểm tra:")
print("A + B =", A + B)
print("C - D =", C - D)
print("A + C =", A + C)
print("B - D =", B - D)
```

#### Nhận xét

Sau 2000 epoch, mô hình tìm được nghiệm xấp xỉ `A = 4.2982`, `B = 4.7017`, `C = 3.7018` và `D = 2.7018`.

Kiểm tra lại bốn phương trình:
- `A + B = 9.0000`
- `C - D = 1.0000`
- `A + C = 8.0000`
- `B - D = 2.0000`

Như vậy, các giá trị tìm được thỏa mãn cả bốn phương trình.

Hệ phương trình có thể có nhiều nghiệm vì bốn phương trình không hoàn toàn độc lập. Từ ba phương trình đầu có thể biểu diễn `B = 9 - A`, `C = 8 - A` và `D = 7 - A`. Khi thay vào phương trình `B - D = 2` thì phương trình này luôn được thỏa mãn. Vì vậy, một biến có thể thay đổi tự do và các biến còn lại thay đổi tương ứng mà vẫn thỏa mãn hệ.

#### Câu 2. Phép toán và thống kê

Tạo hai Tensor `A`, `B` kích thước `3 × 4`; tính `A+B`, `A*B`, `A/B`, trung bình theo cột và độ lệch chuẩn theo cột.

#### Câu 4. Tìm hệ số đa thức

Thay đa thức trong Bài 9 bằng `y = 2x² - 3x + 5`, huấn luyện mô hình và cho biết các hệ số tìm được.

```python
import numpy as np

polynomial = np.poly1d([2, -3, 5])

N = 20
X_data = np.random.randn(N, 1) * 5
Y_data = polynomial(X_data)

XX = np.hstack([
    X_data * X_data,
    X_data,
    np.ones_like(X_data)
])

w = torch.randn(3, 1, requires_grad=True)

x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y_data, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(5000):
    optimizer.zero_grad()

    y_pred = x_train @ w
    mse = torch.mean(torch.square(y_train - y_pred))

    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
```

#### Nhận xét

- **Câu 1:** Tensor `X` gồm 24 phần tử có thể reshape thành các kích thước `4 × 6`, `2 × 12` và `2 × 3 × 4`. Các phép reshape chỉ thay đổi hình dạng của Tensor mà không làm thay đổi tổng số phần tử.

- **Câu 2:** Hai Tensor `A` và `B` có cùng kích thước `3 × 4` nên có thể thực hiện các phép toán theo từng phần tử như cộng, nhân và chia. Khi sử dụng `dim=0`, trung bình và độ lệch chuẩn được tính theo từng cột.

- **Câu 3:** Với hàm `y = 2x² + 5x + 3`, đạo hàm là `dy/dx = 4x + 5`. Tại `x = 2`, ta có `dy/dx = 13`. Kết quả này trùng với gradient được PyTorch tính bằng autograd.

- **Câu 4:** Các hệ số ban đầu được khởi tạo ngẫu nhiên là `[0.5770, -2.1052, 2.1868]`. Sau 5000 epoch, mô hình học được các hệ số xấp xỉ `[2.0000, -3.0000, 4.9999]`, gần như trùng với các hệ số thực `[2, -3, 5]` của đa thức `y = 2x² - 3x + 5`. Điều này cho thấy autograd kết hợp với optimizer NAdam có thể tối ưu các tham số để tìm lại gần chính xác các hệ số của đa thức.
